Venue Analysis Study: Venue Spin Assistance Indexing
Determines venue scoring skew and pitches characteristics.

## Code Implementation
The cell below uses the Midwicket API to calculate metrics and validate cricket analytics findings.

```python
import midwicket as md
from midwicket.features import load_features

def main():
    print("=== Venue Analysis: Spin Assistance ===")
    session = md.datasets.load_dataset("mlc")
    df = load_features("venue_bias_rating", session)
    print(df.head(5))

if __name__ == "__main__":
    main()
```
